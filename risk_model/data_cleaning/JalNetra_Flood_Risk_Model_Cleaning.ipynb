```python
import pandas as pd

df = pd.read_csv("master_dataset.csv")

print("Shape:", df.shape)
df.head()
```

```python
print("Total missing values:", df.isnull().sum().sum())
```

```python
missing_pct = (df.isnull().mean() * 100).sort_values(ascending=False)

print("Missing percentage:")
print(missing_pct[missing_pct > 0])
```

```python
cols_to_inspect = [
    "Location_ID",
    "Location",
    "Road_Network_Unlabeled_Extra_Field",
    "Road_Segment_ID",
    "Road_Length_m",
    "Road_Name",
    "MCG_Unnamed_Col9",
    "MCG hotspot",
    "MCG hotspot listed",
    "MCG documented cause",
    "MCG_Source"
]

inspection = df[cols_to_inspect].copy()

print(inspection.to_string(index=False))
```

```python
print(
    df[
        [
            "Location_ID",
            "Road_Network_Unlabeled_Extra_Field",
            "Road_Segment_ID",
            "Road_Length_m",
            "Nearest_Road_Distance_m",
            "Road_Name"
        ]
    ].to_string(index=False)
)
```

```python
columns_to_drop = [
    "Peak_Rainfall_Time",
    "Drain_Remark",
    "MCG_Unnamed_Col9",
    "Road_Network_Unlabeled_Extra_Field"
]

df_clean = df.drop(columns=columns_to_drop)

print("Original shape:", df.shape)
print("After dropping confirmed unusable columns:", df_clean.shape)

print("\nDropped columns:")
print(columns_to_drop)
```

```python
mcg_cols = [
    "Historical Waterlogging",
    "MCG hotspot listed",
    "MCG documented cause",
    "Verification source",
    "MCG_Source",
    "MCG hotspot"
]

for col in mcg_cols:
    print("\n" + "=" * 70)
    print(f"{col}")
    print("-" * 70)
    print("Missing:", df_clean[col].isna().sum())
    print("Unique values:")
    print(df_clean[col].value_counts(dropna=False).to_string())
```

```python
# Create the modeling target from the canonical location design
df_clean["Flood_Label"] = df_clean["Location_ID"].str.startswith("H").astype(int)

# Columns that directly encode the hotspot/control label
leakage_cols = [
    "Historical Waterlogging",
    "MCG hotspot listed",
    "MCG documented cause",
    "Verification source",
    "MCG_Source",
    "MCG hotspot",
]

# Location_Type, if present, is also direct label information
if "Location_Type" in df_clean.columns:
    leakage_cols.append("Location_Type")

# Remove leakage columns from the modeling dataframe
df_clean = df_clean.drop(columns=leakage_cols, errors="ignore")

print("Flood_Label distribution:")
print(df_clean["Flood_Label"].value_counts().sort_index())

print("\nShape after removing leakage columns:", df_clean.shape)

print("\nRemaining leakage columns:")
print([c for c in leakage_cols if c in df_clean.columns])
```

```python
# Check data types
dtype_summary = pd.DataFrame({
    "Column": df_clean.columns,
    "Data_Type": df_clean.dtypes.astype(str).values,
})

print(dtype_summary.to_string(index=False))

print("\n" + "=" * 70)
print("NON-NUMERIC COLUMNS")
print("=" * 70)

non_numeric = df_clean.select_dtypes(
    exclude=["number"]
).columns.tolist()

print(non_numeric)
print("\nCount:", len(non_numeric))
```

```python
numeric_candidates = [
    "Drain_Size_mm",
    "Road_Length_m",
    "Road_Width_m",
    "Nearest_Road_Distance_m",
    "Drain_Area"
]

for col in numeric_candidates:
    print("\n" + "=" * 70)
    print(f"COLUMN: {col}")
    print("=" * 70)
    print("Unique non-null values:")
    print(df_clean[col].dropna().unique())

    print("\nSample values:")
    print(df_clean[col].dropna().head(10).tolist())
```

```python
import pandas as pd

# Convert straightforward numeric measurements
df_clean["Road_Length_m"] = pd.to_numeric(
    df_clean["Road_Length_m"],
    errors="coerce"
)

df_clean["Road_Width_m"] = pd.to_numeric(
    df_clean["Road_Width_m"],
    errors="coerce"
)

df_clean["Nearest_Road_Distance_m"] = pd.to_numeric(
    df_clean["Nearest_Road_Distance_m"],
    errors="coerce"
)

# Extract numeric value from strings such as:
# "0.841 sq km" -> 0.841
df_clean["Drain_Area"] = (
    df_clean["Drain_Area"]
    .astype(str)
    .str.extract(r"([-+]?\d*\.?\d+)")[0]
    .astype(float)
)

# Verify
cols = [
    "Road_Length_m",
    "Road_Width_m",
    "Nearest_Road_Distance_m",
    "Drain_Area"
]

print(df_clean[cols].dtypes)

print("\nMissing values after conversion:")
print(df_clean[cols].isna().sum())
```

```python
# Keep the original Drain_Size_mm unchanged

size_values = (
    df_clean["Drain_Size_mm"]
    .astype("string")
    .str.findall(r"\d+(?:\.\d+)?")
)

df_clean["Drain_Size_Value_1_mm"] = pd.to_numeric(
    size_values.str[0],
    errors="coerce"
)

df_clean["Drain_Size_Value_2_mm"] = pd.to_numeric(
    size_values.str[1],
    errors="coerce"
)

print(
    df_clean[
        [
            "Location_ID",
            "Drain_Size_mm",
            "Drain_Size_Value_1_mm",
            "Drain_Size_Value_2_mm"
        ]
    ].to_string(index=False)
)

print("\nMissing values:")
print(
    df_clean[
        [
            "Drain_Size_Value_1_mm",
            "Drain_Size_Value_2_mm"
        ]
    ].isna().sum()
)
```

```python
# Inspect remaining object/text columns
text_cols = df_clean.select_dtypes(include=["object", "string"]).columns

for col in text_cols:
    print("\n" + "=" * 70)
    print(f"COLUMN: {col}")
    print(f"Unique values: {df_clean[col].nunique(dropna=True)}")
    print(df_clean[col].dropna().value_counts().head(10).to_string())
```

```python
# ============================================================
# STEP 10 — REMOVE NON-MODEL / LEAKAGE COLUMNS
# ============================================================

drop_cols = [
    # Location identifiers / text
    "Location_ID",
    "Location",

    # Rainfall metadata
    "Rainfall_Source",
    "Rainfall_Data_Quality",

    # Weather metadata / location matching
    "Weather_Query_Time",
    "Matched_Location",
    "Matched_Latitude",
    "Matched_Longitude",
    "Observation_Time",
    "Weather_Source",
    "Weather_Data_Quality",

    # Drainage identifiers / metadata
    "Drain_Name",
    "Drain_Type",
    "Dr_Line_ID",
    "AssetID",
    "FID",
    "Drain_Status",
    "Drain_Source",
    "Drainage_Data_Quality",
    "Drain_Size_mm",

    # Direct / descriptive MCG drainage information
    "MCG_Drainage_Issue",

    # Road identifiers / descriptive metadata
    "Road_Segment_ID",
    "Road_Name",
    "From_Node",
    "To_Node",
    "Road_Source",
    "Road_Data_Quality",
    "Road_Type_Source_Location",
    "Road type / nearest relevant road",

    # Administrative identifiers
    "Ward",
    "Sector",
    "Zone",

    # Traffic metadata
    "Traffic_Source",
    "Traffic_Data_Quality",
    "Time_Observed",

    # Watershed identifiers / metadata
    "Watershed_ID",
    "Natural_Flow_FID",
    "Natural_Flow_Stream_ID",
    "Watershed_Source",
    "Watershed_Data_Quality",

    # Historical incident metadata / leakage
    "Historical Incident Source",
    "Historical Cause"
]

# Drop only columns that actually exist
drop_cols = [col for col in drop_cols if col in df_clean.columns]

df_model = df_clean.drop(columns=drop_cols)

print("Original columns:", len(df_clean.columns))
print("Dropped columns:", len(drop_cols))
print("Remaining columns:", len(df_model.columns))

print("\nRemaining columns:")
print(df_model.columns.tolist())
```

```python
# STEP 11 — Missing values in remaining model columns

model_missing = (
    df_model.isnull()
    .sum()
    .sort_values(ascending=False)
)

model_missing = model_missing[model_missing > 0]

print("Columns with missing values:")
print(model_missing)

print("\nTotal missing values:", df_model.isnull().sum().sum())
```

```python
print("\nCategorical columns and unique values:")

categorical_cols = df_model.select_dtypes(
    include=["object", "string"]
).columns

for col in categorical_cols:
    print(
        f"\n{col} "
        f"({df_model[col].nunique(dropna=True)} unique values):"
    )
    print(df_model[col].dropna().unique())
```

```python
missing_model = df_model.isnull().sum()
missing_model = missing_model[missing_model > 0].sort_values(ascending=False)

print(missing_model.to_string())
```

```python
# STEP 13 — Missingness by Flood_Label

missing_cols = [
    "Drain_Size_Value_2_mm",
    "Road_Length_m",
    "Road_Width_m",
    "Nearest_Road_Distance_m",
    "Drainage_Adequacy",
    "Drainage_Condition",
    "Nearest_Drain_Distance_m",
    "Mapped_Drain_Present",
    "Road_Type",
    "Road_Owner",
    "Road_Maintainer",
    "Drain_Size_Value_1_mm"
]

for col in missing_cols:
    print("\n" + "=" * 65)
    print(col)

    print(
        df_model.groupby("Flood_Label")[col]
        .apply(lambda x: f"{x.isna().sum()}/{len(x)} missing")
        .to_string()
    )
```

```python
coverage_problem_cols = [
    "Drain_Size_Value_1_mm",
    "Drain_Size_Value_2_mm",
    "Road_Length_m",
    "Road_Width_m",
    "Nearest_Road_Distance_m",
    "Drainage_Adequacy",
    "Drainage_Condition",
    "Nearest_Drain_Distance_m",
    "Mapped_Drain_Present",
    "Road_Type",
    "Road_Owner",
    "Road_Maintainer"
]

df_model = df_model.drop(columns=coverage_problem_cols)

print("Shape:", df_model.shape)
print("\nRemaining missing values:")
print(df_model.isnull().sum()[df_model.isnull().sum() > 0])
```

```python
# STEP 15 — Remove columns with only one unique value

constant_cols = [
    col for col in df_model.columns
    if df_model[col].nunique(dropna=False) <= 1
]

print("Constant columns:")
print(constant_cols)

df_model = df_model.drop(columns=constant_cols)

print("\nNew shape:", df_model.shape)
```

```python
print(df_model.columns.tolist())
print("\nData types:")
print(df_model.dtypes)
```

```python
# ============================================================
# STEP 17 — REBUILD FINAL MODEL DATASET CLEANLY
# ============================================================

# Start fresh from df_clean
df_model = df_clean.copy()

# ------------------------------------------------------------
# 1. Remove non-model / metadata / leakage columns
# ------------------------------------------------------------

drop_cols = [
    "Location_ID",
    "Location",

    "Rainfall_Source",
    "Rainfall_Data_Quality",

    "Weather_Query_Time",
    "Matched_Location",
    "Matched_Latitude",
    "Matched_Longitude",
    "Observation_Time",
    "Weather_Source",
    "Weather_Data_Quality",

    "Drain_Name",
    "Drain_Type",
    "Dr_Line_ID",
    "AssetID",
    "FID",
    "Drain_Status",
    "Drain_Source",
    "Drainage_Data_Quality",
    "Drain_Size_mm",
    "MCG_Drainage_Issue",

    "Road_Segment_ID",
    "Road_Name",
    "From_Node",
    "To_Node",
    "Road_Source",
    "Road_Data_Quality",
    "Road_Type_Source_Location",
    "Road type / nearest relevant road",

    "Ward",
    "Sector",
    "Zone",

    "Traffic_Source",
    "Traffic_Data_Quality",
    "Time_Observed",

    "Watershed_ID",
    "Natural_Flow_FID",
    "Natural_Flow_Stream_ID",
    "Watershed_Source",
    "Watershed_Data_Quality",

    "Historical Incident Source",
    "Historical Cause",

    "Peak_Rainfall_Time",
    "Drain_Remark",
    "MCG_Unnamed_Col9",
    "Road_Network_Unlabeled_Extra_Field"
]

df_model = df_model.drop(
    columns=[c for c in drop_cols if c in df_model.columns]
)

# ------------------------------------------------------------
# 2. Remove coverage-biased columns
# ------------------------------------------------------------

coverage_problem_cols = [
    "Drain_Size_Value_1_mm",
    "Drain_Size_Value_2_mm",
    "Road_Length_m",
    "Road_Width_m",
    "Nearest_Road_Distance_m",
    "Drainage_Adequacy",
    "Drainage_Condition",
    "Nearest_Drain_Distance_m",
    "Mapped_Drain_Present",
    "Road_Type",
    "Road_Owner",
    "Road_Maintainer"
]

df_model = df_model.drop(
    columns=[c for c in coverage_problem_cols if c in df_model.columns]
)

# ------------------------------------------------------------
# 3. Remove constant columns
# ------------------------------------------------------------

constant_cols = [
    col for col in df_model.columns
    if df_model[col].nunique(dropna=False) <= 1
]

df_model = df_model.drop(columns=constant_cols)

# ------------------------------------------------------------
# 4. Save Location_ID separately
# ------------------------------------------------------------

location_ids = df_clean["Location_ID"].copy()

# ------------------------------------------------------------
# 5. Standardize column names
# ------------------------------------------------------------

rename_map = {
    "Latitude": "latitude",
    "Longitude": "longitude",
    "Elevation_m": "elevation_m",

    "Rainfall Jun 1–Sep 18 2026 (mm)": "rainfall_2026_mm",
    "Max Hourly Rainfall (mm) 2026": "max_hourly_rainfall_2026_mm",
    "Rainfall 2024 (mm)": "rainfall_2024_mm",
    "Max Hourly Rain 2024 (mm)": "max_hourly_rainfall_2024_mm",
    "Rainfall 2025 (mm)": "rainfall_2025_mm",
    "Max Hourly Rain 2025 (mm)": "max_hourly_rainfall_2025_mm",

    "Route_Distance_m": "route_distance_m",
    "Normal_Duration_sec": "normal_duration_sec",
    "Traffic_Duration_sec": "traffic_duration_sec",
    "Traffic_Delay_sec": "traffic_delay_sec",
    "Congestion_Ratio": "congestion_ratio",
    "Traffic_Avg_Speed_kmh": "traffic_avg_speed_kmh",

    "Distance_to_Watershed_Boundary_m":
        "distance_to_watershed_boundary_m",
    "Natural_Flow_Elevation": "natural_flow_elevation",
    "Natural_Flow_In_Flow": "natural_flow_in_flow",
    "Natural_Flow_Out_Flow": "natural_flow_out_flow",
    "Drain_Area": "drain_area_sq_km",

    "Nearest_Natural_Flow_Distance_m":
        "nearest_natural_flow_distance_m",

    "Traffic_Level": "traffic_level",
    "Flood_Label": "flood_label"
}

df_model = df_model.rename(columns=rename_map)

# ------------------------------------------------------------
# 6. Encode traffic level
# ------------------------------------------------------------

df_model = pd.get_dummies(
    df_model,
    columns=["traffic_level"],
    dtype=int
)

# ------------------------------------------------------------
# 7. Add Location_ID back for tracking
# ------------------------------------------------------------

df_model.insert(
    0,
    "location_id",
    location_ids.values
)

# ------------------------------------------------------------
# 8. Put target at the end
# ------------------------------------------------------------

target = df_model.pop("flood_label")
df_model["flood_label"] = target

# ------------------------------------------------------------
# 9. Final validation
# ------------------------------------------------------------

print("Shape:", df_model.shape)
print("Missing values:", df_model.isnull().sum().sum())
print("Unique locations:", df_model["location_id"].nunique())

print("\nTarget:")
print(df_model["flood_label"].value_counts())

print("\nColumns:")
print(df_model.columns.tolist())

# ------------------------------------------------------------
# 10. Save
# ------------------------------------------------------------

df_model.to_csv(
    "master_model_ready.csv",
    index=False
)

print("\nSaved: master_model_ready.csv")
```
